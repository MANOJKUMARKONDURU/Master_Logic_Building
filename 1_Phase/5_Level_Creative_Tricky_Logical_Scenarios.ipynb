#1. Take coordinates (x, y) and check if the point lies on the X-axis, Y-axis, or at the origin.

```python
def coordinates(x, y):
  if x == 0 and y == 0:
    print("Point is at the origin")
  elif x == 0 and y != 0:
    print('Point lies on the y-axis')
  elif x != 0 and y == 0:
    print('Point lies on the x- axis')
  else:
    print("Point is in quadrents")
x = int(input("Enter the x value: "))
y = int(input("Enter the y value: "))
coordinates(x, y)
```

#2. Take three numbers and check if they can form a Pythagorean triplet.

A set of three numbers forms a Pythagorean triplet if:

(largest number)^2=(other number)^2+(other number)^2

```python
def py_triplet(side1, side2, side3):
  if side2 < side1 > side3:
    largest = side1
    small1 = side2
    small2 = side3
  elif side1 < side2 > side3:
    largest = side2
    small1 = side1
    small2 = side3
  elif side1 < side3 > side2:
    largest = side3
    small1 = side1
    small2 = side2
  if largest ** 2 == small1 ** 2 + small2 ** 2:
      print("Pythagorean Triplet")
  else:
      print("No")

side1 = int(input("Enter the side1: "))
side2 = int(input("Enter the side2: "))
side3 = int(input("Enter the side34: "))
py_triplet(side1, side2, side3)
```

#4. Take time (hours and minutes) and print the smaller angle between the hour and minute hands.
minute_angle = 6 * m

`360˚/60˚minutes=6˚ per minute`

`hour_angle = 30˚ * h + 0.5* m`



1️⃣ Hour hand movement per hour

A full circle (360°) in 12 hours:

`360˚ / 12 hours = 30˚ perhour`

So the hour hand moves:

`30 × ℎ degrees`

2️⃣ Hour hand movement per minute

In one hour (60 minutes), it moves 30°.

So per minute:

`30˚ / 60 = 0.5˚ per minute`

So in 𝑚 minutes, it moves:

0.5 * 𝑚 degrees

Combine both:

`hour_angle =30ℎ + 0.5𝑚`


```python
def angle(h, m):
  #normalize the hours to 12 hours format
  h = h % 12
  minute_angle = 6 * m
  hour_angle = 30 * h + 0.5 * m
  diff = abs(hour_angle - minute_angle)

  smaller_angle = min(diff, 360 - diff)
  print("The smaller angle between the hour and minute hands is ", smaller_angle)

h = int(input("Enter the hour: "))
m = int(input("Enter the minutes: "))
angle(h, m)
```

#5. Take three numbers and check if they are in arithmetic progression.

```python
def arith_prog(n1, n2, n3):
  a, b, c = sorted([n1, n2, n3])

  if b - a == c - b:
    print("In Arithmetic")
  else:
    print("Not in arithmatic")

n1 = int(input("Enter the number1: "))
n2 = int(input("Enter the number2: "))
n3 = int(input("Enter the number3: "))
arith_prog(n1, n2, n3)
```

#6. Take three numbers and check if they are in geometric progression.

```python
def gp(n1, n2, n3):
  a, b, c = sorted([n1, n2, n3])
  r = b/a
  if a * r**2 == c:
    print("In Geometric progression")
  else:
    print("Not In Geometric progression")
n1 = int(input("Enter the number1: "))
n2 = int(input("Enter the number2: "))
n3 = int(input("Enter the number3: "))
gp(n1, n2, n3)
```

```python
def gp(arr):
  arr = sorted(arr)
  if len(arr) == 0:
    print("Not In Geometric progression")
    return
  if all(x == 0 for x in arr):
    print("In Geometric progression")
    return
  if arr[0] == 0:
    print("Not In Geometric progression")
    return
  r = arr[1] / arr[0]
  for i in range(2, len(arr)):
    if arr[i-1] == 0 or arr[i]/arr[i-1] != r:
       print("Not In Geometric progression")
       return
  print("In Geometric progression")
arr = list(map(int, input("Enter the numbers: ").split()))
gp(arr)
```
